```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
dir_path = Path('/scratch/er8/cd3022/xgb_eval/')
df = pd.concat(
    pd.read_csv(f, index_col=0)
    for f in dir_path.glob('*.csv')
)
```

```python
for col in df.columns:
    plt.figure()
    plt.bar(df.index, df[col])
    plt.title(col)
    plt.show()
```

```python
for mod in df.index:
    plt.figure(figsize=(16,5))
    plt.bar(df.columns, df.loc[mod])
    plt.xticks(rotation=45)
    plt.title(mod)
    plt.show()
```
