```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
```

```python
dir_path = Path('/scratch/er8/cd3022/xgb_eval/')
df = pd.concat(
    pd.read_csv(f, index_col=0)
    for f in dir_path.glob('*.csv')
)
```

```python
ax = df.plot(kind='bar')

ax.legend(
    title='Performance metric',
    bbox_to_anchor=(1.05, 1),   # move legend outside
    loc='upper left'
)
ax.set_xlabel('Model')
```

```python
ax = df.T.plot(kind='bar')

ax.legend(
    title='Model',
    bbox_to_anchor=(1.05, 1),   # move legend outside
    loc='upper left'
)

ax.set_xlabel('Performance metric')

plt.show()
```
