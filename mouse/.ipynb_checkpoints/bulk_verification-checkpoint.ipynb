```python
import xgboost as xgb
import xarray as xr
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
from pathlib import Path
import numpy as np

from sklearn.metrics import mean_squared_error
from skimage.metrics import structural_similarity as ssim
```

```python
model = xgb.XGBRegressor()
```

```python
# model.load_model("/scratch/er8/cd3022/xgb_models/test_model.json")
# model.load_model("/scratch/er8/cd3022/xgb_models/radiance_hour_model.json")
```

```python
model.load_model("/scratch/er8/cd3022/xgb_models/alldata_model.json")

test_list = []
data_path = Path('/scratch/er8/cd3022/xgb_datasets/')
for f in data_path.glob('*all_training*'):
    df = pd.read_parquet(f)

    split_index = int(len(df) * 0.8)
    test = df.iloc[split_index:]

    test_list.append(test)


all_test = pd.concat(test_list)

# REMOVE FUTURE VALUES FROM THE TRAINING DATA

X = all_test.drop(columns=['cloud_optical_depth'])
for n in range(1, 7):
    X = X.drop(columns=[f'cloud_optical_depth_t{n}'])

# SELECT FORECAST TIME HORIZON
# (for alldata, times are aligned so have been reduced to BARRA 1hr time steps)
y = all_test['cloud_optical_depth_t1']

y[y == 0] = 0.1
y = np.log(y)

df = X.join(y)
df['cloud_optical_depth'] = all_test['cloud_optical_depth']
df['forecast'] = model.predict(X)

ds_xgb = df.to_xarray()
```

```python
def file_name_to_datetime64(file_path):
    '''
    Convert string from file name into np.datetime64 to be assigned as new coordinate to dataset.
    Assumes filename is in the format ".../YYYYmmddTHHMM..."
    
    INPUTS
    file_path (PosixPath): path to the file

    OUTPUTS
    np.datetime64 corresponding to the file name
    '''
    time_str = file_path.stem
    
    year = time_str[0:4]
    month = time_str[4:6]
    day = time_str[6:8]
    hour = time_str[9:11]
    minute = time_str[11:13]

    return np.datetime64(f'{year}-{month}-{day}T{hour}:{minute}')

def preprocess(ds):
    file_path = Path(ds.encoding["source"])
    t0 = file_name_to_datetime64(file_path)

    # Rename original time → forecast_time
    ds = ds.rename({'time': 'forecast_time'})

    # Convert to lead time (timedelta)
    lead_time = ds['forecast_time'] - t0

    return (
        ds
        .assign_coords(forecast_time=lead_time)
        .expand_dims(time=[t0])
    )
```

```python
gso_files = []
for month in ['01', '03', '06', '09']:
    gso_path = Path(f'/scratch/er8/mm4602/gso/output/nwc_20240701_20251231/2025/{month}/05')
    gso_files.extend([f for f in gso_path.glob('*.nc')])

ds_gso = xr.open_mfdataset(
    gso_files,
    preprocess=preprocess,
    combine='nested',
    concat_dim='time',
    parallel=True,
    chunks='auto'
)
```

```python
ds_gso, ds_xgb = xr.align(
    ds_gso,
    ds_xgb,
    join="inner"
)
```

```python
obs = ds_xgb.cloud_optical_depth
fcst_xgb = ds_xgb.forecast
fcst_gso = ds_gso.sel(forecast_time='01:00:00').cloud_optical_depth
```

```python
mse_xgb = np.mean((fcst_xgb - obs) ** 2)
rmse_xgb = np.sqrt(mse_xgb)
print(f"XGBoost RMSE: {rmse_xgb.item()}")

mse_gso = np.mean((fcst_gso - obs) ** 2)
rmse_gso = np.sqrt(mse_gso)
rmse_gso = rmse_gso.compute()
print(f"Optical Flow RMSE: {rmse_gso.item()}")
```

```python
def compute_ssim(a, b):
    # Handle NaNs
    mask = np.isfinite(a) & np.isfinite(b)
    if mask.sum() == 0:
        return np.nan

    a_valid = a[mask]
    b_valid = b[mask]

    data_range = a_valid.max() - a_valid.min()
    if data_range == 0:
        return np.nan

    # Reshape back to 2D (required by SSIM)
    a2 = np.where(mask, a, 0)
    b2 = np.where(mask, b, 0)

    return ssim(a2, b2, data_range=data_range)
```

```python
ssim_xgb = xr.apply_ufunc(
    compute_ssim,
    obs,
    fcst_xgb,
    input_core_dims=[["latitude", "longitude"], ["latitude", "longitude"]],
    vectorize=True,
    dask="parallelized",
    output_dtypes=[float],
)

ssim_gso = xr.apply_ufunc(
    compute_ssim,
    obs,
    fcst_gso,
    input_core_dims=[["latitude", "longitude"], ["latitude", "longitude"]],
    vectorize=True,
    dask="parallelized",
    output_dtypes=[float],
)
```

```python
print(f"XGBoost SSIM: {ssim_xgb.mean().item()}")
print(f"Optical Flow SSIM: {ssim_gso.mean().compute().item()}")
```
