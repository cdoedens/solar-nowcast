```python
import xgboost as xgb
import xarray as xr
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
from pathlib import Path
import numpy as np
```

```python
model = xgb.XGBRegressor()
```

```python
def transform_cod(y, eps=0.1):
    """
    Log-transform cloud optical depth safely.
    
    Parameters:
        y (array-like): original COD values
        eps (float): small constant to avoid log(0)
    
    Returns:
        transformed y
    """
    return np.log(y + eps)


def inverse_transform_cod(y_log, eps=0.1):
    """
    Inverse transform back to original COD scale.
    """
    return np.exp(y_log) - eps
```

```python
model.load_model("/scratch/er8/cd3022/xgb_models/alldata_model.json")

data_path = Path('/scratch/er8/cd3022/xgb_datasets/')

df  = pd.read_parquet(data_path / 'test.parquet')


X = df.drop(columns=['cloud_optical_depth'])
for n in range(1, 7):
    X = X.drop(columns=[f'cloud_optical_depth_t{n}'])

y = df['cloud_optical_depth_t1']

y_log  = transform_cod(y)

df = X.join(y)
df['cloud_optical_depth_log'] = transform_cod(df['cloud_optical_depth'])
df['forecast_log'] = model.predict(X)

ds_xgb = df.to_xarray()
```

```python
def file_name_to_datetime64(file_path):
    '''
    Convert string from file name into np.datetime64 to be assigned as new coordinate to dataset.
    Assumes filename is in the format ".../YYYYmmddTHHMM..."
    
    INPUTS
    file_path (PosixPath): path to the file

    OUTPUTS
    np.datetime64 corresponding to the file name
    '''
    time_str = file_path.stem
    
    year = time_str[0:4]
    month = time_str[4:6]
    day = time_str[6:8]
    hour = time_str[9:11]
    minute = time_str[11:13]

    return np.datetime64(f'{year}-{month}-{day}T{hour}:{minute}')

def preprocess(ds):
    # Extract filename from encoding
    file_path = Path(ds.encoding["source"])
    
    # Add new "time" dimension
    t = file_name_to_datetime64(file_path)
    
    return (
        ds
        .rename({'time': 'forecast_time'})
        .expand_dims(time=[t])
        .sel(
            latitude=ds_xgb.latitude,
            longitude=ds_xgb.longitude
        )
    )
```

```python
gso_files = []
for month in ['01', '03', '06', '09']:
    gso_path = Path(f'/scratch/er8/mm4602/gso/output/nwc_20240701_20251231/2025/{month}/05')
    gso_files.extend([f for f in gso_path.glob('*.nc')])

ds_gso = xr.open_mfdataset(
    gso_files,
    preprocess=preprocess,
    combine='nested',
    concat_dim='time',
    parallel=True,
    chunks='auto'
)
```

```python
ds_gso['cloud_optical_depth'] = xr.where(ds_gso['cloud_optical_depth'] == 0, 0.1, ds_gso['cloud_optical_depth'])
ds_gso['cloud_optical_depth'] = np.log(ds_gso.cloud_optical_depth)
```

```python
fig, ax = plt.subplots(nrows=1, ncols=4, figsize=(16,10))
time= '2025-01-05T02:00'
f_time= '2025-01-05T03:00'

vmin=0
vmax=np.log(200)


im0 = ax[0].imshow(
    ds_xgb.cloud_optical_depth.sel(time=time),
    vmin=vmin, vmax=vmax
)
ax[1].imshow(
    ds_xgb.cloud_optical_depth_t1.sel(time=time),
    vmin=vmin, vmax=vmax
)

ax[2].imshow(
    ds_gso.cloud_optical_depth.sel(time=time, forecast_time=f_time),
    vmin=vmin, vmax=vmax
)

ax[3].imshow(
    ds_xgb.forecast.sel(time=time),
    vmin=vmin, vmax=vmax
)

ax[0].set_title('observation')
ax[1].set_title('observation t60min')
ax[2].set_title('GSO forecast t60min')
ax[3].set_title('XGB forecast t60min')

cbar = fig.colorbar(im0, ax=ax, orientation='vertical', fraction=0.02, pad=0.04)
cbar.set_label('Cloud Optical Depth (log scale)')
```
