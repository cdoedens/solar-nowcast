# Evaluation

- Evaluate training and test data distributions
- Display feature importances
- Inspect SHAP results

```python
import xgboost as xgb
import pandas as pd
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt

import shap

import sys

sys.path.append('/home/548/cd3022/repos/solar-nowcast/modules')
import data_transform
```

```python
model = xgb.XGBRegressor()
```

```python
model.load_model("/scratch/er8/cd3022/xgb_models/alldata_model.json")
```

```python
data_path = Path('/scratch/er8/cd3022/xgb_datasets/')

train_df = pd.read_parquet(data_path / 'train.parquet')
test_df  = pd.read_parquet(data_path / 'test.parquet')
```

```python
X_train = train_df.drop(columns=['cloud_optical_depth'])
X_test = test_df.drop(columns=['cloud_optical_depth'])
for n in range(1, 7):
    X_train = X_train.drop(columns=[f'cloud_optical_depth_t{n}'])
    X_test = X_test.drop(columns=[f'cloud_optical_depth_t{n}'])

y_train = train_df['cloud_optical_depth_t1']
y_test = test_df['cloud_optical_depth_t1']

y_train_log = transform_cod(y_train)
y_test_log  = transform_cod(y_test)
```

```python
y_pred_train = model.predict(X_train)
y_pred_test = model.predict(X_test)
```

# Data Distributions

```python
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(6, 6))
ax[0,0].hist(y_train_log, bins=100)
ax[0,0].set_title("True-train")
ax[0,0].set_xlim(-3, 5)
ax[0,0].set_ylim(0, 800_000)

ax[0,1].hist(y_pred_train, bins=100)
ax[0,1].set_title("Prediction-train")
ax[0,1].set_xlim(-3, 5)
ax[0,1].set_ylim(0, 800_000)

ax[1,0].hist(y_test_log, bins=100)
ax[1,0].set_title("True-test")
ax[1,0].set_xlim(-3, 5)
ax[1,0].set_ylim(0, 800_000)

ax[1,1].hist(y_pred_test, bins=100)
ax[1,1].set_title("Prediction-test")
ax[1,1].set_xlim(-3, 5)
ax[1,1].set_ylim(0, 800_000)
```

```python
y_true = y_train_log
y_pred = y_pred_train

plt.figure()

plt.hist2d(
    y_true,
    y_pred,
    bins=100,
    cmap='viridis',
    vmax=70_000
)

plt.colorbar(label='Counts')

# 1:1 line
lims = [min(y_true.min(), y_pred.min()),
        max(y_true.max(), y_pred.max())]

plt.plot(lims, lims, 'r--')

plt.xlabel("True values")
plt.ylabel("Predicted values")
plt.title("Predicted vs True (Heatmap density)")
plt.show()
```

```python
y_true = y_test_log
y_pred = y_pred_test

plt.figure()

plt.hist2d(
    y_true,
    y_pred,
    bins=100,
    cmap='viridis',
    vmax=70_000
)

plt.colorbar(label='Counts')

# 1:1 line
lims = [min(y_true.min(), y_pred.min()),
        max(y_true.max(), y_pred.max())]

plt.plot(lims, lims, 'r--')

plt.xlabel("True values")
plt.ylabel("Predicted values")
plt.title("Predicted vs True (Heatmap density)")
plt.show()
```

# SHAP Results

```python
explainer = shap.Explainer(model)

idx = np.random.choice(len(X_test), size=20000, replace=False)
X_sample = X_test.iloc[idx]

shap_values = explainer(X_sample)
```

```python
shap.plots.bar(shap_values)
```
