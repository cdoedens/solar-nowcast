```python
from pathlib import Path
from datetime import datetime, timedelta
import matplotlib.pyplot as plt
import xarray as xr
import numpy as np
import pandas as pd

import os
import sys

sys.path.append('/home/548/cd3022/repos/solar-nowcast/modules')
import sat_preprocess

import pyearthtools.data as petdata
import pyearthtools.pipeline as petpipe
from pyearthtools.data.time import Petdt
from pyearthtools.pipeline.operations.xarray.join import GeospatialTimeSeriesMerge
import site_archive_nci
```

```python
year = 2024
month = 1
day = 1
lat_min=-35
lat_max=-32
lon_min=148
lon_max=152
```

```python
file_path = Path(f'/g/data/rv74/satellite-products/arc/der/himawari-ahi/solar/p1s/latest/{year}/{month:02d}/{day:02d}')
files = [f for f in file_path.rglob('*.nc')]

def preprocess(ds):
    return ds.sel(
        latitude=slice(lat_min, lat_max),
        longitude=slice(lon_min, lon_max)
    )[['surface_global_irradiance', 'solar_elevation']]

himawari = xr.open_mfdataset(files, preprocess=preprocess)
```

```python
files = []
variables_of_interest = [
    'huss',
    'hus850',
    'hus700',
    'hus500',
    'psl',
    'tas',
    'ta850',
    'ta700',
    'ta500',
]
for var in variables_of_interest:
    file_path = Path(f'/g/data/ob53/BARRA2/output/reanalysis/AUST-04/BOM/ERA5/historical/hres/BARRA-C2/v1/1hr/{var}/latest/')
    var_file = [f for f in file_path.glob(f'*{year}{month:02d}.nc')][0]
    files.append(var_file)

def preprocess(ds):
    return ds.sel(
        lat=slice(lat_min, lat_max),
        lon=slice(lon_min, lon_max)
    )
bar = xr.open_mfdataset(
    files,
    compat='override',
    preprocess=preprocess
)
```

```python
# ds_him = himawari.sel(latitude=-33.84,longitude=151.22, method='nearest')
# ds_bar = bar.sel(lat=-33.84,lon=151.22, method='nearest')

bar_interp = bar.interp(
    lat=himawari.latitude,
    lon=himawari.longitude,
    method='nearest'
)

bar_aligned, himawari_aligned = xr.align(
    bar_interp,
    himawari,
    join="inner"
)

ds = xr.merge([bar_aligned, himawari_aligned]).drop_vars(['crs', 'sigma', 'height', 'level_height', 'model_level_number', 'pressure', 'lat', 'lon'])

for n in range(1, 11):
    ds[f'surface_global_irradiance_t{n}'] = (
        ds['surface_global_irradiance'].shift(time=-n)
    )
```

```python
df = ds.to_dataframe()
data = df.dropna()
```
