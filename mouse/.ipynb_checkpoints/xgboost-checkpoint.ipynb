```python
import xgboost as xgb

from pathlib import Path
from datetime import datetime, timedelta
import matplotlib.pyplot as plt
import xarray as xr
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import os

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
data_path = Path('/scratch/er8/cd3022/xgb_datasets/')
df = pd.concat(
    pd.read_parquet(f)
    for f in data_path.glob('*all_training*')
)
df['month'] = df.index.get_level_values('time').month
```

```python
test_months = [2, 6, 10]

train_df = df[~df['month'].isin(test_months)]
test_df  = df[df['month'].isin(test_months)]
```

```python
train_df.to_parquet(data_path / 'train.parquet')
test_df.to_parquet(data_path / 'test.parquet')
```

```python
X_train = train_df.drop(columns=['cloud_optical_depth'])
X_test = test_df.drop(columns=['cloud_optical_depth'])
for n in range(1, 7):
    X_train = X_train.drop(columns=[f'cloud_optical_depth_t{n}'])
    X_test = X_test.drop(columns=[f'cloud_optical_depth_t{n}'])
y_train = train_df['cloud_optical_depth_t1']
y_test = test_df['cloud_optical_depth_t1']
```

```python
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(6, 3))
ax[0].hist(y_train, bins=100)
ax[0].set_title("Original")

ax[1].hist(np.log(y_train + 0.1), bins=100)
ax[1].set_title("Log-transformed")
```

```python
def transform_cod(y, eps=0.1):
    """
    Log-transform cloud optical depth safely.
    
    Parameters:
        y (array-like): original COD values
        eps (float): small constant to avoid log(0)
    
    Returns:
        transformed y
    """
    return np.log(y + eps)


def inverse_transform_cod(y_log, eps=1e-6):
    """
    Inverse transform back to original COD scale.
    """
    return np.exp(y_log) - eps
```

```python
y_train_log = transform_cod(y_train)
y_test_log  = transform_cod(y_test)
```

```python
model = xgb.XGBRegressor(
    random_state=42,
    n_estimators=2000,
    early_stopping_rounds=50,
    learning_rate=0.03,
    eval_metric='rmse',
    # device='cuda'
)

model.fit(
    X_train, y_train_log,
    eval_set=[(X_train, y_train_log), (X_test, y_test_log)],
    verbose=True
)

# Predict on test set
y_pred = model.predict(X_test)

# Quick evaluation of model performance using correlation and RMSE
correlation = np.corrcoef(y_test_log, y_pred)[0, 1]
rmse = np.sqrt(mean_squared_error(y_test_log, y_pred))

print(f"Correlation between predicted and actual values: {correlation:.3f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.3f}")
```

```python
model.save_model("/scratch/er8/cd3022/xgb_models/alldata_model.json")
```

# Validation

# Cross-Validation

```python
months = sorted(df['month'].unique())

results = []
models = {}

for test_month in months:
    print(f"\n=== Holding out month {test_month} ===")
    
    train_df = df[df['month'] != test_month]
    test_df  = df[df['month'] == test_month]

    # Split features/target
    X_train = train_df.drop(columns=['cloud_optical_depth'])
    X_test  = test_df.drop(columns=['cloud_optical_depth'])

    # remove lagged COD features (your existing logic)
    for n in range(1, 7):
        col = f'cloud_optical_depth_t{n}'
        if col in X_train.columns:
            X_train = X_train.drop(columns=[col])
            X_test = X_test.drop(columns=[col])

    y_train = train_df['cloud_optical_depth_t1']
    y_test  = test_df['cloud_optical_depth_t1']

    # log transform target
    y_train_log = np.log(y_train + 1e-6)
    y_test_log  = np.log(y_test + 1e-6)

    model = xgb.XGBRegressor(
        random_state=42,
        n_estimators=300,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric='rmse'
    )

    model.fit(
        X_train, y_train_log,
        eval_set=[(X_train, y_train_log), (X_test, y_test_log)],
        verbose=False
    )

    y_pred = model.predict(X_test)

    rmse = np.sqrt(mean_squared_error(y_test_log, y_pred))

    results.append((test_month, rmse))
    models[test_month] = model
```

```python
cv_results = pd.DataFrame(results, columns=['month', 'rmse'])
print(cv_results)

print("\nMean RMSE:", cv_results['rmse'].mean())
print("Std RMSE:", cv_results['rmse'].std())
```

```python
results = models[test_month].evals_result()

train_loss = results['validation_0']['rmse']
val_loss = results['validation_1']['rmse']

plt.figure()
plt.plot(train_loss, label='Train RMSE')
plt.plot(val_loss, label='Validation RMSE')

plt.xlabel('Boosting Iterations')
plt.ylabel('RMSE')
plt.title('Training vs Validation Loss')
plt.legend()

plt.show()
```
